### Configuración común.

Esta celda es **idéntica en todos los *notebooks***: lee la configuración de variables de entorno (o del archivo `.env`, si existe) y usa valores por omisión cuando no están definidas. Muestra todas las variables. **En este *notebook* se comprueba que todas se leen bien.** Las claves solo se muestran como «definida / no definida». La tabla completa de variables está en el `README.md` y en `00_preliminares_entorno_y_secretos.ipynb`.

````python
# Configuración común de py521 (celda idéntica en todos los notebooks).
import os
from pathlib import Path

from dotenv import find_dotenv, load_dotenv

load_dotenv(find_dotenv(usecwd=True), override=False)  # las variables del sistema tienen prioridad sobre .env

def _entorno(nombre, por_omision=None):
    valor = os.environ.get(nombre, "").strip()
    return valor or por_omision

# Servicio compatible con OpenAI (Ollama, LM Studio, llama.cpp server, vLLM...).
LLM_BASE_URL = _entorno("LLM_BASE_URL", _entorno("LLM_DEFAULT_BASE_URL", "http://localhost:11434/v1")).rstrip("/")  # el contenedor fija LLM_DEFAULT_BASE_URL
LLM_API_KEY = _entorno("LLM_API_KEY")              # secreto: sin valor por omisión
LLM_MODEL = _entorno("LLM_MODEL", "qwen3:1.7b")
LLM_EMBED_MODEL = _entorno("LLM_EMBED_MODEL", "qwen3-embedding:0.6b")
LLM_MODEL_A = _entorno("LLM_MODEL_A", "qwen3:1.7b-q4_K_M")
LLM_MODEL_B = _entorno("LLM_MODEL_B", "qwen3:1.7b-q8_0")
LLM_CONTEXT_A = int(_entorno("LLM_CONTEXT_A", "4096"))
LLM_CONTEXT_B = int(_entorno("LLM_CONTEXT_B", "16384"))
RAG_DOCS_DIR = Path(_entorno("RAG_DOCS_DIR", "data"))
LLM_PROXY_KEY = _entorno("LLM_PROXY_KEY")          # secreto de la demo 05: sin valor por omisión
LLM_PROXY_HOST = _entorno("LLM_PROXY_HOST", "127.0.0.1")
LLM_PROXY_PORT = int(_entorno("LLM_PROXY_PORT", "8800"))

# API nativa de Ollama (solo para las funciones marcadas «específico de Ollama»).
# Se usa OLLAMA_HOST, como el cliente oficial de Ollama; si no existe, se deriva de LLM_BASE_URL.
def _url_ollama(valor):
    valor = (valor or LLM_BASE_URL.removesuffix("/v1")).strip().rstrip("/")
    if "://" not in valor:
        anfitrion = valor.split("/")[0]
        valor = "http://" + (valor if ":" in anfitrion.split("]")[-1] else valor.replace(anfitrion, anfitrion + ":11434", 1))
    return valor.replace("://0.0.0.0", "://localhost")  # 0.0.0.0 es una dirección de escucha, no de destino

OLLAMA_URL = _url_ollama(_entorno("OLLAMA_HOST"))

for nombre in ["LLM_BASE_URL", "LLM_MODEL", "LLM_EMBED_MODEL", "LLM_MODEL_A", "LLM_MODEL_B",
               "LLM_CONTEXT_A", "LLM_CONTEXT_B", "RAG_DOCS_DIR", "LLM_PROXY_HOST", "LLM_PROXY_PORT", "OLLAMA_URL"]:
    print(f"{nombre:<16} {globals()[nombre]}")
for nombre in ["LLM_API_KEY", "LLM_PROXY_KEY", "HF_TOKEN"]:
    print(f"{nombre:<16} {'definida' if _entorno(nombre) else 'no definida'}")
````

### 1.8 Comprobación del servicio.

* La siguiente celda buscará el programa `ollama`, comprobará que el servicio de `LLM_BASE_URL` responde y listará sus modelos.
* Resultado esperado: «Servicio en …: responde ✅» y, en la celda posterior, ✅ en cada modelo de las demos.

````python
print("Programa ollama:", shutil.which("ollama") or "no está en el PATH (no es necesario si el servicio responde o usas otro motor)")

from openai import OpenAI

cliente = OpenAI(base_url=LLM_BASE_URL, api_key=LLM_API_KEY or "sin-clave")  # Ollama y LM Studio sin autenticación ignoran la clave

try:
    disponibles = sorted(m.id for m in cliente.models.list().data)
    print(f"Servicio en {LLM_BASE_URL}: responde ✅")
except Exception as error:
    disponibles = []
    print(f"Servicio en {LLM_BASE_URL}: no responde ❌ ({type(error).__name__}). Abre Ollama o revisa LLM_BASE_URL.")
````

* La siguiente celda mostrará los modelos que ofrece el servicio. Con *Ollama* añade el tamaño, la cuantización y si el modelo es local o de nube (🦙 específico de *Ollama*).

````python
import httpx

def ollama_detalles():
    """🦙 Tamaño, cuantización y tipo (local o nube) de los modelos, según /api/tags de Ollama."""
    try:
        return {m["name"]: m for m in httpx.get(f"{OLLAMA_URL}/api/tags", timeout=10).json()["models"]}
    except Exception:
        print("ℹ️ El servicio no es Ollama: se muestran solo los nombres.")
        return {}

detalles = ollama_detalles()
print(f"{'Modelo':<32}{'Tamaño':>10}  {'Cuantización':<14}Tipo")
for nombre in disponibles:
    m = detalles.get(nombre, {})
    nube = "remote_host" in m or nombre.endswith("cloud")
    tamano = f"{m['size'] / 1e9:.1f} GB" if m and not nube else "—"
    cuant = m.get("details", {}).get("quantization_level", "—") if m else "—"
    print(f"{nombre:<32}{tamano:>10}  {cuant:<14}{('nube' if nube else 'local') if m else '—'}")

print()
for modelo in dict.fromkeys([LLM_MODEL, LLM_MODEL_A, LLM_MODEL_B, LLM_EMBED_MODEL]):
    print(f"{modelo:<24}", "✅" if modelo in disponibles else f"❌ no disponible (en Ollama: ollama pull {modelo})")
````

## 2. Secretos.

### 2.1 Qué es un secreto.

Un **secreto** es un valor que da acceso a un servicio o a datos: claves de API, *tokens* de acceso, contraseñas o claves privadas. Quien conoce el valor puede actuar con los permisos de su dueño.

* **Nunca** se escribe en una celda del *notebook*: el código y las salidas se guardan en el archivo `.ipynb` y se comparten con él.
* **Nunca** se sube al repositorio: aunque se borre después, queda en el historial de *git* y en las copias que otros ya descargaron.
* Se guarda fuera del código, en **variables de entorno** o en un archivo `.env` que no se sube.

### 2.2 Variables de entorno y archivo `.env`.

* Una **variable de entorno** es un valor que el sistema operativo entrega a los programas al iniciarlos. *Python* la lee con `os.environ`.
* El archivo `.env` guarda variables de entorno en líneas `NOMBRE=valor`. La biblioteca [*python-dotenv*](https://pypi.org/project/python-dotenv/) lo lee y carga sus valores en `os.environ`.
* Este repositorio incluye `.env.example` con los nombres de las variables y **sin valores**. Para usarlo, cópialo como `.env` y escribe los valores en tu equipo:

```bash
cp .env.example .env    # En Windows: copy .env.example .env
```

* `.env` está en `.gitignore`, así que *git* no lo sube.

La celda de configuración común (sección 1.7) ya cargó `.env` si existe; las variables ya definidas en el sistema tienen prioridad (`override=False`).

````python
print("Archivo .env:", "existe" if find_dotenv(usecwd=True) else "no existe (se usan solo las variables del sistema y los valores por omisión)")
````

### 2.3 Leer un secreto sin imprimirlo.

* Se comprueba **si está definido**, nunca se muestra su valor, ni completo ni en parte.
* No se imprime `os.environ` completo: contiene todos los secretos del entorno.
* Al compartir un *notebook*, se borran antes las salidas (*Kernel → Restart Kernel and Clear Outputs*).

````python
SECRETOS = {
    "HF_TOKEN": "Token de Hugging Face (solo para modelos con acceso restringido)",
    "LLM_API_KEY": "Clave que se envía al servicio, si la pide",
    "LLM_PROXY_KEY": "Clave que exige el proxy de la demo 05",
}

def estado_secreto(nombre):
    """Indica si una variable de entorno tiene valor sin mostrarlo."""
    return "definido" if os.environ.get(nombre) else "no definido"

for nombre, descripcion in SECRETOS.items():
    print(f"{nombre:<15} {estado_secreto(nombre):<12} {descripcion}")
````

* Cuando una biblioteca necesita el secreto, se le pasa la variable, no el texto: por ejemplo, `token=os.environ["HF_TOKEN"]`.

### 2.4 Comprobar que `.env` no se sube al repositorio.

* La siguiente celda preguntará a *git* si `.env` está ignorado. Si la carpeta no es un repositorio de *git*, lo indicará.

````python
import subprocess

if shutil.which("git"):
    resultado = subprocess.run(["git", "check-ignore", "-q", ".env"], capture_output=True)
    if resultado.returncode == 0:
        print(".env está ignorado por git ✅")
    elif resultado.returncode == 1:
        print(".env NO está ignorado ❌: añade la línea `.env` a .gitignore")
    else:
        print("Esta carpeta no es un repositorio de git; la comprobación no aplica.")
else:
    print("git no está instalado; la comprobación no aplica.")
````

### 2.5 Qué hacer si un secreto se filtra.

Un secreto se considera filtrado si apareció en un *commit*, en una salida guardada, en una captura de pantalla o en un mensaje.

1. **Revocar:** invalidarlo de inmediato en el servicio que lo emitió. Borrarlo del archivo o del historial **no** basta.
2. **Rotar:** crear un secreto nuevo con los permisos mínimos y actualizar `.env` en cada equipo que lo usa.
3. **Revisar:** consultar en el servicio si hubo actividad no reconocida mientras estuvo expuesto.
4. **Limpiar:** quitarlo del archivo, de las salidas y, si llegó a *git*, del historial; esto se hace después de revocar.

* *Hugging Face*: el *token* se elimina o se renueva en <https://huggingface.co/settings/tokens> ([guía oficial](https://huggingface.co/docs/hub/security-tokens#revoking-a-leaked-token)).
* *GitHub* detecta algunos secretos conocidos en los repositorios con el [escaneo de secretos](https://docs.github.com/es/code-security/secret-scanning/introduction/about-secret-scanning); no sustituye a la revocación.
* La clave del proxy de la demo 05 se rota generando una nueva (sección 3.2) y reiniciando el proxy.

Las demos de este repositorio **no necesitan ninguna credencial externa**: los modelos de *Ollama* se descargan sin cuenta. Esta sección sirve para cuando haga falta.

### 3.1 *Token* de *Hugging Face* para modelos con acceso restringido.

Algunos modelos de *Hugging Face* tienen **acceso restringido** (*gated*): hay que iniciar sesión, aceptar sus condiciones en la página del modelo y descargarlos con un *token* ([documentación oficial](https://huggingface.co/docs/hub/models-gated)).

1. Inicia sesión y entra a <https://huggingface.co/settings/tokens> → **New token**.
2. Elige el tipo con el **mínimo permiso** ([tipos de *token*](https://huggingface.co/docs/hub/security-tokens)):
   * **Fine-grained** (recomendado): permisos por recurso; marca solo la lectura de los repositorios con acceso restringido que vas a usar.
   * **Read**: solo lectura de todo lo que tu cuenta puede leer.
   * **Write** no hace falta para descargar modelos; no lo uses.
3. Usa un *token* por equipo o por uso, con un nombre que lo identifique (por ejemplo, `laptop-py521`).
4. **Caducidad:** la documentación oficial no describe una fecha de caducidad para estos *tokens*; elimínalo en la misma página cuando termines de usarlo.
5. Copia el valor en `.env` como `HF_TOKEN=...`. No lo pegues en una celda ni en la terminal.

### 3.2 Clave para limitar el acceso al servicio (demo 05).

La demo 05 coloca un proxy con clave delante de *Ollama*, porque la API de *Ollama* no pide autenticación. La clave la genera la propia persona: no la emite ningún servicio externo.

* Debe ser larga y aleatoria; se genera con el módulo `secrets` de *Python*.
* Se guarda en `.env` del equipo que ejecuta el proxy como `LLM_PROXY_KEY`. Cada aplicación cliente la guarda en su propio `.env` como `LLM_API_KEY` y la envía en la cabecera `Authorization: Bearer ...`.
* **Caducidad:** la decide quien administra el servicio; para la demo, se rota al terminar la sesión.
* Si se usa *LM Studio* como servicio, la clave se crea en **Developer → Server Settings → Require Authentication → Manage Tokens**, con los permisos mínimos; el valor se muestra una sola vez ([documentación oficial](https://lmstudio.ai/docs/developer/core/authentication)).

La siguiente celda **crea** la clave y la guarda en `.env` sin mostrarla, solo si cambias `CREAR_CLAVE` a `True` y la clave aún no existe.

````python
import secrets
from pathlib import Path

CREAR_CLAVE = False  # cambia a True para generar LLM_PROXY_KEY en .env

def crear_clave_en_env(nombre="LLM_PROXY_KEY", archivo=".env"):
    """Genera una clave aleatoria y la añade a .env sin mostrarla."""
    if os.environ.get(nombre):
        print(f"{nombre} ya está definido; no se modifica.")
        return
    ruta = Path(archivo)
    with ruta.open("a", encoding="utf-8") as f:
        f.write(f"\n{nombre}={secrets.token_urlsafe(32)}\n")
    load_dotenv(ruta, override=False)  # load_dotenv se importó en la configuración común
    print(f"{nombre} creado en {ruta.resolve().name}: {estado_secreto(nombre)}.")

if CREAR_CLAVE:
    crear_clave_en_env()
else:
    print(f"LLM_PROXY_KEY: {estado_secreto('LLM_PROXY_KEY')}. Para crearla, cambia CREAR_CLAVE a True.")
````
